```python
import numpy as np
import matplotlib.pyplot as plt
%matplotlib qt
```

```python
def display(A):
    maxX = A.shape[0]
    maxY = A.shape[1]
    B = np.zeros((maxY, maxX))
    for ix in range(0,maxX):
        for iy in range(0,maxY):
            B[maxY-1-iy,ix] = A[ix,iy]

    %matplotlib qt
    
    plt.rcParams['figure.figsize'] = [6, 6/maxX*maxY]
    plt.imshow(B); 
    plt.axis('off'); 
    plt.show()
    plt.draw()
    plt.pause(0.01)
```

```python
nParticles = 70000
maxX = 200
maxY = 40
A = np.zeros((maxX, maxY))
A[:,0] = 1
A[:, -1] = 1 

yStart  = maxY//2 
nFrames = 50
B = np.zeros((nFrames, maxX, maxY))
success = 0; failure = 0
```

```python
for i in range(0,nParticles):
    x  = maxX-1
    y  = yStart
    steps = 0

    while True:
        xOrg = x
        yOrg = y

        r = np.random.random(); 
        if 0.0 <= r < .5:
            x = x-1
        if .5 <= r < .6:
            x = x+1
        if .6 <= r < .8:
            y = y+1
        if .8 <= r < 1.0:
            y = y-1

        steps += 1

        if x <= 0 or x == 0:
            success += 1
            break

        if x > (maxX-1):
            x = xOrg; y = yOrg
            continue

        if y < 0 or y > (maxY-1):
            failure += 1
            break
        xp = (x+1)%maxX 
        xm = (x-1)%maxX 
        yp = (y+1)%maxY
        ym = (y-1)%maxY

        if (A[xp,y] == 1) or (A[xm,y] == 1) or (A[x,yp] == 1) or (A[x,ym] == 1):
            failure += 1
            break

        if (steps % 10) == 0:
            f = steps // 10
            if 1 <= f <= nFrames and (0 <= x < maxX) and (0 <= y < maxY):
                B[f-1, x, y] += 1

        
        if xm < 0: xm = xm+maxX
        if xp > (maxX-1): xp = 0

    if (i+1) % 1000 == 0:
        p  = success / (i+1)
        se = (p*(1-p)/(i+1))**0.5
        print(f"After {i+1} particles: p(success)={p:.4f}, SE≈{se:.4f}")
        if se < 0.002:
            print("Accuracy is less than 0.2% at ")
            break
```

```python
print(f"Success probability is {p:.4f} at less than 0.2% accuracy error at {i} Particles")
```

```python
def display_2(A,B):
    maxX = A.shape[0]
    maxY = A.shape[1]

    m2 = np.max(B)
    B[A[:,:]==1] = m2 
    if (m2>0):
        m1 = np.min(B[B[:,:]>0])
        if (m1==m2): m1 = m2/2
        B[B[:,:]==0] = m1*(m1/m2)**(1/10) 

    D  = np.zeros((maxY, maxX))
    for iy in range(0,maxY):
        D[maxY-1-iy,:] = np.log(B[:,iy])

    plt.imshow(D);
    plt.axis('off');
    plt.show()
    plt.draw()
    plt.pause(0.1) 
```

```python
while(True):
    for i in range(nFrames):
        display_2(A, B[i,:,:])
```
