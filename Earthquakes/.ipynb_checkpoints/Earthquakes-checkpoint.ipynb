```python
minLat = 33.6
maxLat = 37.0
minLon = -103.0
maxLon = -94.6

t1 = op.UTCDateTime("2015-12-01")
t2 = op.UTCDateTime("2016-02-01")
minMag = 2.0

catalog = client.get_events(starttime=t1, endtime=t2,
                            minlatitude=minLat, maxlatitude=maxLat,
                            minlongitude=minLon, maxlongitude=maxLon,
                            minmagnitude=minMag)

print("Number of earthquakes:", len(catalog))

days = []

for i in range(len(catalog)):
    days.append((catalog[i].origins[0].time - t1) / 86400.0)  # days since Dec 1

plt.hist(days, bins=50)
plt.xlabel("Days Since Dec 1, 2015")
plt.ylabel("Number of Earthquakes")
plt.show()

swarm = []

for i in range(len(catalog)):
    lat1 = catalog[i].origins[0].latitude
    lon1 = catalog[i].origins[0].longitude
    t1   = catalog[i].origins[0].time
    
    group = []
    
    for j in range(i+1, len(catalog)):
        t2 = catalog[j].origins[0].time
        dt = t2 - t1   # time difference in seconds
        
        if dt > 14*24*3600:
            break
        
        lat2 = catalog[j].origins[0].latitude
        lon2 = catalog[j].origins[0].longitude
        
        dlat = (lat2 - lat1)*111.0
        dlon = (lon2 - lon1)*111.0 * math.cos((lat1+lat2)/2 * 3.14/180)
        dist = (dlat**2 + dlon**2)**0.5
        
        if dist < 10:
            group.append(catalog[j])
    
    if len(group) >= 40:
        swarm = group
        break

print("Swarm size:", len(swarm))
```

```python
all_lat = []
all_lon = []
for i in range(len(catalog)):
    all_lat.append(catalog[i].origins[0].latitude)
    all_lon.append(catalog[i].origins[0].longitude)

sw_lat = []
sw_lon = []
for ev in swarm:
    sw_lat.append(ev.origins[0].latitude)
    sw_lon.append(ev.origins[0].longitude)

fig = plt.figure(figsize=(50,30))
ax = fig.add_subplot(1, 1, 1, projection=ccrs.PlateCarree())
ax.set_extent([minLon, maxLon, minLat, maxLat], crs=ccrs.PlateCarree())
ax.gridlines()

import cartopy.io.img_tiles as cimg
request = cimg.GoogleTiles()
ax.add_feature(cfeature.BORDERS, linestyle=':')
ax.add_image(request, 8)

ax.scatter(all_lon, all_lat, c='gray', s=200, transform=ccrs.PlateCarree())
ax.scatter(sw_lon, sw_lat, c='red', s=400, transform=ccrs.PlateCarree())
```
