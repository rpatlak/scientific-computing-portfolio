```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
n = 11
data = np.zeros((n,n))
for i in range(n):
    data[i:, i] = 1          
plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
n = 21
r = 10
c = n//2
data = np.zeros((n,n))
for i in range(n):
    if i == 10:
        data[i:, i] = 1 
        data[:i, i] = 1 
        data[i, :i] = 1
        data[i, i:] = 1

for j in range (r+1):
    data[c-r+j, c-j]=1
    data[c-r+j, c+j]=1
    data[c-r-j-1, c-j]=1
    data[c-r-j-1, c+j]=1

plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
n = 51
c==n/2
data = np.zeros((n,n))
for i in range(n):
    data[i, i:n-i] = 1          
    data[i, n - i-1 :i+1] = 1
    data[c,c]==1
plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
n = 11
data = np.zeros((n,n))
for i in range(n):
    for j in range(n):
        if (i+j)%2==0:
            data[i,j]=1
plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
n = 13
data = np.zeros((n,n))
for i in range(n):
    for j in range(n):
        if (i+2*j)%3==0:
            data[i,j]=1
    
plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
n = 501
c = n // 2
r = c

data = np.zeros((n, n))
for i in range(n):
    for j in range(n):
        x = abs(i - c) / r
        y = abs(j - c) / r
        if x**(1/2) + y**(1/2) > 1: 
            data[i, j] = 1

plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
n = 501
data = np.zeros((n, n))

cy = n// 2           
cx1, cx2 = 0.35*n, 0.65*n  
r = 0.35*n
r2 = r*r

for i in range(n):       
    for j in range(n):
        circle1 = ((i - cy)**2) + ((j - cx1)**2) <= r2
        circle2 = ((i - cy)**2) + ((j - cx2)**2) <= r2
        if ((circle1 + circle2) % 2):
            data[i, j] = 1 
        else:
            data[i, j] =0

plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = 1001
data = np.ones((n, n))   

cx, cy = 500, 500        
R = 450                
half_thick = 7          
r_in2  = (R - half_thick) ** 2
r_out2 = (R + half_thick) ** 2

for i in range(n):              
    for j in range(n):         
        dx, dy = j - cx, i - cy
        d2 = dx ** 2 + dy ** 2
        if r_in2 <= d2 <= r_out2:
            data[i, j] = 0

eye_r = 75
eye_r2 = eye_r ** 2
ex1, ey1 = 350, 380
ex2, ey2 = 650, 380

for i in range(n):              
    for j in range(n):    
        dx1, dy1 = j - ex1, i - ey1
        if dx1 ** 2 + dy1 ** 2 <= eye_r2:
            data[i, j] = 0

for i in range(n):              
    for j in range(n):           
        dx2, dy2 = j - ex2, i - ey2
        if dx2 ** 2 + dy2 ** 2 <= eye_r2:
            data[i, j] = 0
            
mx, my = 500, 500   
mr = 360          
y_cut = 700       
mr2 = mr ** 2

for i in range(n):              
    for j in range(n):    
        dxm, dym = j - mx, i - my
        if (dxm ** 2 + dym** 2 <= mr2) and (i >= y_cut):
            data[i, j] = 0

plt.imshow(data, interpolation='nearest')
plt.show()
```

```python
H = np.loadtxt('topography_180x360_grid.txt')

ny, nx = H.shape

H2 = np.copy(H)
for ix in range(nx):
    for iy in range(ny):
        ix2 = (ix + 200) % nx                     
        H2[iy, ix] = H[iy, ix2]

H3 = np.copy(H2)
for ix in range(nx):
    for iy in range(ny):
        ix2 = int(round((nx-1)/2 + (ix - (nx-1)/2)/2.0))  
        if ix2 < 0: ix2 = 0
        if ix2 > nx-1: ix2 = nx-1
        H3[iy, ix] = H2[iy, ix2]

plt.imshow(H3, interpolation='nearest')
plt.show()
```
