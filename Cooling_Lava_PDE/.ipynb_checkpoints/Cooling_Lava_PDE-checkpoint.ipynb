```python
import numpy as np

N = 50
L = 13 * N
dx = 10
dt = 900
k = 0.042
c = 1.23
p = 2.55
iterations = int(31536000 // dt) + 1

x = np.linspace(0, L * dx, L + 1)
T = np.zeros(L)

center = (L * dx) / 2
half_dike = 250

for i in range(L):
    if (center - half_dike) <= x[i] <= (center + half_dike):
        T[i] = 1150
    else:
        T[i] = 20

it = 0
time = 0.0
center_idx = L // 2

E_initial = 0
for i in range(L - 1):
    if (center - half_dike) <= x[i] <= (center + half_dike):
        E_initial += 0.5 * (T[i] + T[i + 1]) * (x[i + 1] - x[i])

for it in range(iterations):
    T_new = np.copy(T)
    for i in range(1, L - 1):
        T_new[i] = T[i] + (dt * k / (c * p * dx**2)) * (T[i + 1] - 2 * T[i] + T[i - 1])

    T_new[0] = 20
    T_new[-1] = 20

    T = np.copy(T_new)
    time = it * dt

    E_now = 0
    for i in range(L - 1):
        if (center - half_dike) <= x[i] <= (center + half_dike):
            E_now += 0.5 * (T[i] + T[i + 1]) * (x[i + 1] - x[i])

    fraction = E_now / E_initial

    if abs(time - 3600) < dt:
        print("1 hour:", T[center_idx], "Fraction:", fraction)
    elif abs(time - 86400) < dt / 2:
        print("1 day:", T[center_idx], "Fraction:", fraction)
    elif abs(time - 604800) < dt / 2:
        print("1 week:", T[center_idx], "Fraction:", fraction)
    elif abs(time - 2592000) < dt / 2:
        print("1 month:", T[center_idx], "Fraction:", fraction)
    elif abs(time - 31536000) < dt:
        print("1 year:", T[center_idx], "Fraction:", fraction)
```
