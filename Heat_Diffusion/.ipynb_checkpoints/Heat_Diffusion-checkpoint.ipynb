```python
from matplotlib.animation import FFMpegWriter
from matplotlib import cm

N = 51
T = np.zeros([N,N])

T[:, 0]  = 0.0
T[:, -1] = 0.0
T[0, :]  = 0.0
T[-1, :] = 0.0

T[7,  5:-5]  = 100    
T[-7, 5:-5]  = 100   

mid = N//2
T[5:-4, mid-1:mid+2] = 100

iterations = 2000
T_new = T.copy()
from mpl_toolkits.mplot3d import Axes3D
from matplotlib import cm
X = np.linspace(0, L, N) 
Y = np.linspace(0, L, N)
X, Y = np.meshgrid(X, Y)
fig = plt.figure(dpi=200)

ax = fig.add_subplot(projection='3d')
ax.set_zlim(0, 100)

filename = "Homework 7 Question 1.mp4"
metadata = dict(title="3D Heat Diffusion (I shape)", artist="Matplotlib")
writer = FFMpegWriter(fps=15, metadata=metadata)

with writer.saving(fig, filename, dpi=200):
    for n in range(iterations):
        for i in range(1, N-1):
            for j in range(1, N-1):
                T_new[i, j] = 0.25 * (T[i-1, j] + T[i+1, j] + T[i, j-1] + T[i, j+1])
        
        T[:] = T_new[:]

        ax.clear()
        ax.set_zlim(0, 100)
        ax.plot_surface(X, Y, T, cmap=cm.coolwarm, antialiased=False)
        ax.set_title(f"Iteration {n+1}")

        writer.grab_frame()

        plt.pause(0.001)

print(f"Saved MP4: {filename}")
```

```python
from matplotlib.animation import FFMpegWriter
from matplotlib import cm
T2 = np.zeros([N, N])
N=51
mid = N//2
iterations = 2000


T2[:, 0]  = 0.0
T2[:, -1] = 0.0
T2[0, :]  = 0.0
T2[-1, :] = 0.0

T2[7,  5:-5]  = 100    
T2[-7, 5:-5]  = 100   
T2[5:-4, mid-1:mid+2] = 100

T2[7,  5:-5]  = 100    
T2[-7, 5:-5]  = 100   


T2_new = T2.copy()

fig2 = plt.figure(dpi=200)
ax2 = fig2.add_subplot(projection='3d')
ax2.set_zlim(0, 100)

filename2 = "Homework 7 Question 2 (insulated).mp4"
metadata2 = dict(title="3D Heat Diffusion (Insulated BCs)", artist="Matplotlib")
writer2 = FFMpegWriter(fps=15, metadata=metadata2)

energy = np.zeros(iterations)
energy[0] = T2.sum()

with writer2.saving(fig2, filename2, dpi=200):
    for n in range(iterations):
        # interior points
        for i in range(1, N-1):
            for j in range(1, N-1):
                T2_new[i, j] = 0.25 * (T2[i-1, j] + T2[i+1, j] + T2[i, j-1] + T2[i, j+1])

        for i in range(1, N-1):
            j = 0
            T2_new[i, j] = 0.25 * (T2[i, j] + T2[i+1, j] + T2[i-1, j] + T2[i, j+1])
            j = N-1
            T2_new[i, j] = 0.25 * (T2[i, j] + T2[i+1, j] + T2[i-1, j] + T2[i, j-1])

        for j in range(1, N-1):
            i = 0
            T2_new[i, j] = 0.25 * (T2[i, j] + T2[i+1, j] + T2[i, j-1] + T2[i, j+1])
            i = N-1
            T2_new[i, j] = 0.25 * (T2[i, j] + T2[i-1, j] + T2[i, j-1] + T2[i, j+1])

        T2_new[0, 0]         = 0.25 * (2.0*T2[0,0]     + T2[1,0]     + T2[0,1])
        T2_new[0, N-1]       = 0.25 * (2.0*T2[0,N-1]   + T2[1,N-1]   + T2[0,N-2])
        T2_new[N-1, 0]       = 0.25 * (2.0*T2[N-1,0]   + T2[N-2,0]   + T2[N-1,1])
        T2_new[N-1, N-1]     = 0.25 * (2.0*T2[N-1,N-1] + T2[N-2,N-1] + T2[N-1,N-2])

        T2[:] = T2_new[:]

        energy[n] = T2.sum()


        NX, NY = T2.shape
        X2 = np.linspace(0, L, NX)
        Y2 = np.linspace(0, L, NY)
        X2, Y2 = np.meshgrid(X2, Y2)

        ax2.clear()
        ax2.set_zlim(0, 100)
        ax2.plot_surface(X2, Y2, T2, cmap=cm.coolwarm, antialiased=False)
        ax2.set_title(f"Insulated BCs — Iter {n+1}, Energy = {energy[n]:.2f}")

        writer2.grab_frame()
        plt.pause(0.001)

print(f"Saved MP4: {filename2}")

emin = energy.min()
emax = energy.max()
drift = (emax - emin) / (abs(energy[0]) + 1e-12)
print(f"Energy values: min={emin:.6f}, max={emax:.6f}, relative drift = {drift:.6e}")
```

```python
import numpy as np

# Grid and parameters
N = 51
L = 30.0

T = np.zeros((N, N))      # temperature field
T_new = T.copy()

iterations = 2000

# Boundary conditions (fixed edges)
T[:, 0]  = 0.0
T[:, -1] = 0.0
T[0, :]  = 0.0
T[-1, :] = 0.0

# Initial condition
mid = N // 2
T[7,  5:-5]  = 100
T[-7, 5:-5]  = 100
T[5:-4, mid-1:mid+2] = 100

# Time stepping
for n in range(iterations):
    # Interior points
    # T_new = average of four neighbors
    # Discrete Laplacian diffusion
    for i in range(1, N-1):
        for j in range(1, N-1):
            T_new[i, j] = 0.25 * (
                T[i-1, j] +
                T[i+1, j] +
                T[i, j-1] +
                T[i, j+1]
            )
    # Left and right boundaries
    # Zero-flux (insulated): mirror value across boundary
    for i in range(1, N-1):
        j = 0
        T_new[i, j] = 0.25 * (
            T[i, j] +
            T[i+1, j] +
            T[i-1, j] +
            T[i, j+1]
        )
        j = N-1
        T_new[i, j] = 0.25 * (
            T[i, j] +
            T[i+1, j] +
            T[i-1, j] +
            T[i, j-1]
        )
    # Top and bottom boundaries
    for j in range(1, N-1):
        i = 0
        T_new[i, j] = 0.25 * (
            T[i, j] +
            T[i+1, j] +
            T[i, j-1] +
            T[i, j+1]
        )
        i = N-1
        T_new[i, j] = 0.25 * (
            T[i, j] +
            T[i-1, j] +
            T[i, j-1] +
            T[i, j+1]
        )
    # Corners
    # Two mirrored neighbors → 2.0*T term
    T_new[0, 0] = 0.25 * (
        2.0*T[0, 0] +
        T[1, 0] +
        T[0, 1]
    )
    T_new[0, N-1] = 0.25 * (
        2.0*T[0, N-1] +
        T[1, N-1] +
        T[0, N-2]
    )
    T_new[N-1, 0] = 0.25 * (
        2.0*T[N-1, 0] +
        T[N-2, 0] +
        T[N-1, 1]
    )
    T_new[N-1, N-1] = 0.25 * (
        2.0*T[N-1, N-1] +
        T[N-2, N-1] +
        T[N-1, N-2]
    )
    # Update temperature field
    T[:] = T_new[:]
```
