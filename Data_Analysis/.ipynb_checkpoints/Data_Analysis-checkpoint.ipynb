```python
import matplotlib.pyplot as plt
import numpy as np
import glob
%config InlineBackend.figure_format = 'retina'
```

```python
self_guess = np.loadtxt("X_Guesses_Patlak_Riley.txt")
solutions = np.loadtxt("x_positions_2025.txt")
```

```python
nNumbers = len(self_guess)
print(nNumbers)
```

```python
plt.plot(self_guess, linestyle = "--", label="Self Guess")
plt.plot(solutions, linestyle = "--", label="Solutions")

plt.xlabel("X positions")
plt.ylabel("Value")
plt.legend()
plt.show()
```

```python
plt.scatter(solutions, self_guess)
```

```python
difference = solutions - self_guess
plt.plot(difference)
plt.axhline(y=0, color="orange")
print(f"the most positive deviation was {np.max(difference)}")
print(f"the most negative deviation was {np.min(difference)}")
```

```python
average = (np.sum(difference) / 100)
print(average)
```

```python
std_dev = np.sqrt((np.sum(difference ** 2))/nNumbers) 
print(f"the standard deviation is {std_dev}")
print(f"the variance is {std_dev**2}")
```

```python
diff = difference
plt.hist(diff, 50)
```

```python
filenames = glob.glob('submissions_2025/*.txt')
```

```python
plt.rcParams['figure.figsize'] = [15, 7]
for file in filenames:
    data = np.loadtxt(file);
    #print('File loaded: ',file,'\n length of dataset=',data.shape[0],'   minimum and maximum values= ',np.min(data),np.max(data));
    #print(data)
    plt.plot(data,'k-');

plt.show()
```

```python
solutions = np.loadtxt("x_positions_2025.txt")

plt.rcParams['figure.figsize'] = [15, 7]
for file in filenames:
   data = np.loadtxt(file);
   plt.scatter(solutions, data)

x = [0,1]
y = [0,1]
plt.plot(x,y)
```

```python
plt.rcParams['figure.figsize'] = [15, 7]
solutions = np.loadtxt("x_positions_2025.txt")

for file in filenames:
    data = np.loadtxt(file)
    deviation = data - solutions
    plt.plot(deviation, label=file)

plt.xlabel("Index (0–99)")
plt.ylabel("Deviation from solution")
plt.show()

print("Most negative deviation =", np.min([np.min(np.loadtxt(f) - solutions) for f in filenames]))
print("Most positive deviation =", np.max([np.max(np.loadtxt(f) - solutions) for f in filenames]))
```

```python
data = np.loadtxt(filenames[0])
difference = data - solutions
nNumbers = len(difference)
std_dev = np.sqrt((np.sum(difference**2))/nNumbers)
print("Standard deviation for first dataset =", std_dev)

stdDev = np.zeros(len(filenames))
for i,file in enumerate(filenames):
    data = np.loadtxt(file)
    difference = data - solutions
    nNumbers = len(difference)
    stdDev[i] = np.sqrt((np.sum(difference**2))/nNumbers)
print("Standard deviations for all datasets:")
print(stdDev)
```

```python
sorted_stdDev = np.sort(stdDev)
plt.plot(sorted_stdDev)
print(sorted_stdDev)
```

```python
solutions = np.loadtxt("x_positions_2025.txt")

diffAll = []
for file in filenames:
    data = np.loadtxt(file)
    difference = data - solutions
    for d in difference:
        diffAll.append(d)

plt.hist(diffAll, bins=100)  
plt.xlim(-0.25,0.25) 
plt.show()
```

```python
nNumbers = len(diffAll)
sigma = np.sqrt(np.sum(np.array(diffAll)**2)/nNumbers)
print("sigma =", sigma)
```
