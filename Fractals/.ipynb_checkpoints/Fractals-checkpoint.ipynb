```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def Mandelbrot(z, C):
    zNew = z**2 + C
    return zNew
```

```python
nMax = 300
N = 1001
x = np.linspace(-1.7, 1.7, N)   
y = np.linspace(-1.7, 1.7, N)
A = np.zeros((N, N))

C = -0.4 + 0.6j
zMax = 50

for i in range(N):
    for j in range(N):
        z = x[i] + 1j * y[j]
        for n in range(nMax):
            z = Mandelbrot(z, C)
            if np.abs(z) >= zMax:
                A[j, i] = n
                break
        else:
            A[j, i] = nMax
plt.title("Part 1")
plt.imshow(A)
```
